# Descriptive Analysis of the Student Survey

This notebook looks at data summaries to understand the main variables and outcomes.

```python
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data_path = Path("../1_data/2_clean/student_survey_cleaned.dta")
df = pd.read_stata(data_path)
```

### Helper Functions

```python
def frequency_table(data, variable):
    """
    Produce counts and valid percentages for one categorical variable.
    Missing observations are excluded from the percentage denominator.
    """

    series = data[variable]
    valid = series.dropna()

    table = (
        valid.value_counts(sort=False)
        .rename_axis("Category")
        .reset_index(name = "Frequency")
    )
    

    table["Percent"] = (
        table["Frequency"]
        .div(len(valid))
        .mul(100)
        .round(1)
    )
    print(
        f"{variable}: "
        f"Total N = {len(series)}, "
        f"Valid N = {series.notna().sum()}, "
        f"Missing = {series.isna().sum()}"
    )
    return table

    
```

```python
age_summary = df["age"].describe(percentiles=[0.25, 0.5, 0.75]).rename({
    "count": "Valid N", "mean": "Mean", "std": "SD", "min": "Minimum",
    "25%": "25th percentile", "50%": "Median", "75%": "75th percentile", "max": "Maximum",
}).to_frame("Age").round(1)
display(age_summary)
```

```python
demographic_variables = {
    "Gender": "gender",
    "Education level and year": "education",
    "Province": "Province",
    "Local government": "lg_name",
    "Local-government type": "lg_type",
    "Urban/rural upbringing": "life_place",
    "Ethnicity": "ethnicity",
    "Religious belief": "religious_belief",
    "Religion": "religion",
    "Marital status": "marital_status",
}
```

### Descriptive table for demographic variables

```python
from IPython.display import display

for title, variable in demographic_variables.items():
    print(f"\n{title}")
    display(frequency_table(df, variable))
```
